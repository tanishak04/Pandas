Merging and Joining 

```python
import pandas as pd
```

```python
data=pd.read_csv(r"C:\Users\mohan\python\engineering-log\pandas\datasets\orders.csv")
data1=pd.read_csv(r"C:\Users\mohan\python\engineering-log\pandas\datasets\products.csv")
```

```python
print("Data before operation\n\n",data ,"\n\n", data1)
```

```python
merge=pd.merge(data,data1, on="ProductID", how="left")
```

```python
print(merge)
```

```python
merge=pd.merge(data,data1,on="ProductID",how="right")
print(merge)
```
