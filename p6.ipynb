INTERPOLATION

interpolate estimates missing numerical values based on the surrounding values instead of simply using the mean/median.

```python
import pandas as pnd
```

```python
data=pnd.read_csv(r"C:\Users\mohan\python\engineering-log\pandas\datasets\missing.csv")
```

```python
print(data)
```

```python
data["Salary"]=data["Salary"].interpolate(method="linear")
```

```python
print(data)
```

Polynomial Interpolation

```python
data["Salary"]=data["Salary"].interpolate(method="polynomial",order=2)
print(data)
```
