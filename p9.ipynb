## Grouping in Pandas

`groupby()` groups rows based on one or more columns and allows calculations to be performed separately on each group.

**Syntax:**
`data.groupby("group_column")["column"].operation()`

Example:
`data.groupby("Department")["Salary"].mean()` → Calculates the average salary for each department.

````python
import pandas as pd
````

````python
data=pd.read_csv(r"C:\Users\mohan\python\engineering-log\pandas\datasets\missing.csv")
````

````python
print("Data before operation\n\n",data)
````

After grouping

````python
group=data.groupby("Department")["Salary"].mean()
````

````python
print(group)
````

````python
group=data.groupby("Department")["Salary"].mean()
````

````python
group=data.groupby(["Experience","City"])["Age"].mean()
````

## Common Aggregation Functions

Aggregation functions summarize multiple values into a single result.

| Function | Description |
|---|---|
| `mean()` | Calculates the average |
| `sum()` | Calculates the total |
| `min()` | Finds the smallest value |
| `max()` | Finds the largest value |
| `count()` | Counts non-missing values |
| `median()` | Finds the middle value |
| `std()` | Calculates standard deviation |
| `var()` | Calculates variance |
| `first()` | Returns the first value |
| `last()` | Returns the last value |
